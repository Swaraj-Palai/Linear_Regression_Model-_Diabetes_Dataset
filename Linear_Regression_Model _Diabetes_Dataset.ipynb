```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score 
from sklearn import datasets
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
db=datasets.load_diabetes()
X=db.data
y=db.target
print(X.shape)
print(type(X))
print(y.shape)
print(db.DESCR)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
print(X_train.shape)
print(X[0:5], y[0:5])
print(y[0:5])
```

```python
fig, axes = plt.subplots(nrows=2, ncols=5, figsize=(15,8), sharey=True)

axes = axes.flatten()

for i in range(10):
    axes[i].scatter(X[:, i], y, alpha=0.5, s=5)
    axes[i].set_title(f"Feature {i+1} vs Target")
    axes[i].set_xlabel(f"Feature {i+1}")
    
    if i % 5 == 0:
        axes[i].set_ylabel("Target y")

plt.tight_layout()
plt.show()
```

```python
model = LinearRegression()
print(model)
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
print(y_pred)
```

```python
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print(f"Mean squared error: {mse}")
print(f"Coefficient of determination (R2 score) : {r2}")
```

```python
print("No of features seen during fit: ", model.n_features_in_)
```

```python
print("model coefficients:",model.coef_)
print("model Intercept:",model.intercept_)
```
